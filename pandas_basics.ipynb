# Pandas Basics Tutorial for Freshers

Pandas is a powerful Python library for data manipulation and analysis. This notebook covers essential functions and concepts for beginners.

---

## 1. Installation & Import

Before using pandas, we need to import it. It's commonly imported as `pd`.

```python3
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

## 2. Main Data Structures

Pandas has two main data structures:
- **Series**: 1D array with labeled indices
- **DataFrame**: 2D table with labeled rows and columns

### 2.1 Series - 1D Data Structure

```python3
# Creating a Series from a list
scores = pd.Series([85, 90, 78, 92, 88])
print("Series from list:")
print(scores)
print()

# Creating a Series with custom index
student_scores = pd.Series([85, 90, 78, 92, 88], 
                            index=['Alice', 'Bob', 'Charlie', 'David', 'Eve'])
print("Series with custom index:")
print(student_scores)
```

### 2.2 DataFrame - 2D Data Structure

```python3
# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 28, 35, 32],
    'Score': [85, 90, 78, 92, 88],
    'Department': ['HR', 'IT', 'Sales', 'IT', 'HR']
}

df = pd.DataFrame(data)
print(df)
```

## 3. Basic DataFrame Information & Exploration

These functions help you understand your data quickly.

```python3
# Display first few rows (default 5)
print("First 3 rows:")
print(df.head(3))
print()

# Display last few rows
print("Last 2 rows:")
print(df.tail(2))
```

```python3
# Get shape (rows, columns)
print(f"Shape: {df.shape}")
print()

# Get column names
print(f"Column names: {df.columns.tolist()}")
print()

# Get data types of each column
print("Data types:")
print(df.dtypes)
```

```python3
# Get basic info about the DataFrame
df.info()
```

```python3
# Get statistical summary
print("Statistical summary:")
print(df.describe())
```

## 4. Accessing Data

Learn how to select specific rows, columns, or cells from a DataFrame.

```python3
# Access a single column
print("Accessing 'Name' column:")
print(df['Name'])
print()

# Alternative way using dot notation (works for simple column names)
print("Using dot notation:")
print(df.Name)
```

```python3
# Access multiple columns
print("Accessing multiple columns:")
print(df[['Name', 'Age', 'Score']])
```

```python3
# Access rows using iloc (integer location)
print("Row at index 0 using iloc:")
print(df.iloc[0])
print()

# Access rows using loc (label-based)
print("Using loc to access by condition:")
print(df.loc[df['Age'] > 28])
```

```python3
# Access a specific cell
print(f"Cell at row 2, column 'Name': {df.loc[2, 'Name']}")
print(f"Cell at row 0, column 1 (using iloc): {df.iloc[0, 1]}")
```

## 5. Filtering & Conditional Selection

Select rows based on conditions.

```python3
# Filter rows where Age > 28
print("Employees with Age > 28:")
filtered = df[df['Age'] > 28]
print(filtered)
print()

# Multiple conditions
print("Employees with Age > 28 AND Score > 85:")
filtered2 = df[(df['Age'] > 28) & (df['Score'] > 85)]
print(filtered2)
```

```python3
# Filter by string value
print("Employees in IT department:")
it_dept = df[df['Department'] == 'IT']
print(it_dept)
print()

# Using isin() for multiple values
print("Employees in HR or Sales:")
print(df[df['Department'].isin(['HR', 'Sales'])])
```

## 6. Adding & Modifying Columns

```python3
# Add a new column
df['Bonus'] = df['Score'] * 0.1
print("DataFrame with Bonus column:")
print(df)
print()

# Create column based on condition
df['Performance'] = df['Score'].apply(lambda x: 'High' if x >= 85 else 'Low')
print("\nDataFrame with Performance column:")
print(df)
```

```python3
# Modify existing column
df['Age'] = df['Age'] + 1  # Increase all ages by 1
print("Updated DataFrame:")
print(df)
```

## 7. Removing Rows & Columns

```python3
# Drop a column
df_copy = df.copy()
df_copy = df_copy.drop('Bonus', axis=1)  # axis=1 means column
print("DataFrame after dropping 'Bonus' column:")
print(df_copy)
print()

# Drop rows by index
df_copy2 = df.copy()
df_copy2 = df_copy2.drop(0, axis=0)  # axis=0 means row
print("\nDataFrame after dropping row 0:")
print(df_copy2)
```

## 8. Sorting Data

```python3
# Sort by a single column
print("Sorted by Age (ascending):")
print(df.sort_values('Age'))
print()

# Sort by column in descending order
print("Sorted by Score (descending):")
print(df.sort_values('Score', ascending=False))
```

```python3
# Sort by multiple columns
print("Sorted by Department then by Score:")
print(df.sort_values(['Department', 'Score'], ascending=[True, False]))
```

## 9. Aggregation & Grouping

Summarize data using group operations.

```python3
# Basic aggregation functions
print("Sum of Scores:", df['Score'].sum())
print("Average Age:", df['Age'].mean())
print("Maximum Score:", df['Score'].max())
print("Minimum Score:", df['Score'].min())
print("Count of entries:", len(df))
print("Standard deviation:", df['Score'].std())
```

```python3
# GroupBy - group data by a column
print("Group by Department:")
grouped = df.groupby('Department')
print(grouped.head())
```

```python3
# Aggregate grouped data
print("Average Score by Department:")
print(df.groupby('Department')['Score'].mean())
print()

print("Count of employees by Department:")
print(df.groupby('Department').size())
```

```python3
# Multiple aggregations
print("Multiple stats by Department:")
agg_stats = df.groupby('Department').agg({
    'Age': ['mean', 'min', 'max'],
    'Score': ['mean', 'count']
})
print(agg_stats)
```

## 10. Handling Missing Data

```python3
# Create data with missing values
data_with_missing = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Score': [85, None, 78, 92, 88]
}
df_missing = pd.DataFrame(data_with_missing)
print("DataFrame with missing values:")
print(df_missing)
```

```python3
# Check for missing values
print("Missing values count:")
print(df_missing.isnull().sum())
print()

# Find rows with any missing values
print("Rows with missing values:")
print(df_missing[df_missing.isnull().any(axis=1)])
```

```python3
# Drop rows with missing values
print("After dropping missing values:")
print(df_missing.dropna())
print()

# Fill missing values
print("After filling missing values with 0:")
print(df_missing.fillna(0))
```

## 11. Data Types & Conversion

```python3
# Create sample data
data_types_df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': ['25', '30', '28'],  # String, should be int
    'Score': [85.5, 90.2, 78.9],
    'Is_Manager': [True, False, True]
})

print("Original data types:")
print(data_types_df.dtypes)
print()

# Convert Age to integer
data_types_df['Age'] = pd.to_numeric(data_types_df['Age'])
print("After converting Age to numeric:")
print(data_types_df.dtypes)
```

## 12. String Operations

Manipulate string columns using `.str` accessor.

```python3
# Create sample data
names_df = pd.DataFrame({
    'Name': ['alice smith', 'bob jones', 'charlie brown']
})

print("Original:")
print(names_df)
print()

# Convert to uppercase
print("Uppercase:")
print(names_df['Name'].str.upper())
print()

# Get string length
print("String length:")
print(names_df['Name'].str.len())
print()

# Extract first name
print("First name (before space):")
print(names_df['Name'].str.split(' ').str[0])
```

## 13. Merging & Joining DataFrames

```python3
# Create two DataFrames to merge
employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Dept_ID': [10, 20, 10, 30]
})

departments = pd.DataFrame({
    'Dept_ID': [10, 20, 30],
    'Dept_Name': ['HR', 'IT', 'Sales']
})

print("Employees:")
print(employees)
print("\nDepartments:")
print(departments)
```

```python3
# Merge DataFrames
merged = pd.merge(employees, departments, on='Dept_ID')
print("Merged DataFrame:")
print(merged)
```

```python3
# Concatenate DataFrames
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

concatenated = pd.concat([df1, df2], ignore_index=True)
print("Concatenated:")
print(concatenated)
```

## 15. Pivot Tables

Reshape data for analysis.

```python3
# Create sample data
sales_data = pd.DataFrame({
    'Region': ['North', 'South', 'North', 'South', 'East', 'East'],
    'Product': ['A', 'B', 'B', 'A', 'A', 'B'],
    'Sales': [100, 150, 200, 120, 180, 160]
})

print("Original data:")
print(sales_data)
print()

# Create pivot table
pivot = sales_data.pivot_table(values='Sales', index='Region', columns='Product', aggfunc='sum')
print("Pivot table:")
print(pivot)
```

## Practice Exercises

Try these exercises to practice pandas!

```python3
# Exercise 1: Create a simple DataFrame
# Create a DataFrame with student data and find the average score

student_data = {
    'Name': ['Student1', 'Student2', 'Student3', 'Student4'],
    'Math': [85, 90, 78, 92],
    'Science': [88, 85, 80, 95],
    'English': [92, 88, 85, 89]
}

df_students = pd.DataFrame(student_data)
print("Student DataFrame:")
print(df_students)
print()

# Calculate average score per student
df_students['Average'] = df_students[['Math', 'Science', 'English']].mean(axis=1)
print("\nWith Average Score:")
print(df_students)
```

```python3
# Exercise 2: Filter and sort
# Find students with average > 85 and sort by average descending

top_students = df_students[df_students['Average'] > 85].sort_values('Average', ascending=False)
print("Top students (Average > 85):")
print(top_students)
```

```python3
# Exercise 3: Apply functions
# Create a grade column based on average

def assign_grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 80:
        return 'B'
    else:
        return 'C'

df_students['Grade'] = df_students['Average'].apply(assign_grade)
print("Students with Grades:")
print(df_students)
```
